Для упражнений в курсах по тестированию постоянно создаются функции

- чтения файла
- получения пути к нему 
- фикстура конечного результата
- tmp_path - это временный каталог - точное имя не нужно передавать, либа сама понимает, что от нее нужно (как бы)
- `dest = tmp_path / file_name` = так конструируется путь+имя нового файла 
- по новому путь+имя файла копируется файл с тестовыми данными
- уже по новому путь+имя берется файл, передается в тестируемую функцию и результат сравнивается "как обычно"
